```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
tip_data=pd.read_excel('/content/tip-amount.xlsx')
tip_data.head(10)

```

```python
tip_data.drop('Meal',axis=1,inplace=True)
```

```python
tip_data.corr()
```

```python
#feature- target splitting
y=tip_data['Observed tip amount(yi)']
x=tip_data.drop('Observed tip amount(yi)',axis=1)
```

```python
# train - test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)

x_test
```

```python
#creating linear regression model
from sklearn.linear_model import LinearRegression
lr=LinearRegression()

model=lr.fit(x_train,y_train)
y_pred=model.predict(x_test)
print("Pred :",y_pred)
print(y_test)
```

```python
from sklearn.metrics import mean_squared_error,r2_score
print("MSE:",mean_squared_error(y_test,y_pred))
print("R2:",r2_score(y_test,y_pred))
```

```python
import pandas as pd

# MAKING PREDICTION FOR 50$
while True:
    input_bill = int(input("\nEnter Bill Amount : "))
    bill_pred = model.predict(pd.DataFrame([[input_bill]], columns=['Observed total bill(xi)']))

    pred_df = pd.DataFrame({'Observed total bill': [input_bill], 'Predicted Tip Amount': bill_pred.flatten()})
    display(pred_df)
```
